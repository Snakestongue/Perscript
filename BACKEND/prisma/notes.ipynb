Generator
1. Prisma-Client: A library allowing JS to connect with Postgre SQL
2. Output: Tells Prisma where to put the code.

Datasource: Allows Prisma to know what language I'm using. I will use Postgre SQL

Models: Allows Prisma to learn the different taqbles in my code. A lot of this will be designed sort of as I go.

Keywords
@id: Primary key
@unique: No duplicates allowed
@default(cuid()): Automatically generates an ID (super helpful)
String: Text
DateTime: Date and time :/
@relation: Connects two tables
@updatedAt: Automatically updates when a record changes
